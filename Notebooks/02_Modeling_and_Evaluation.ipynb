# Modeling and Evaluation

## 1. Load Data and Define Features and Target

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path("..").resolve()))
from Helper.helper import predict_rul
```

```python
import pandas as pd


df = pd.read_csv("../Data/Cleaned/train_FD001_cleaned.csv")

X = df.drop(columns=["RUL", "engine_id"])


y = df["RUL"]


groups = df["engine_id"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("Number of engines:", groups.nunique())

display(X.head())
```

## 2. Define Models and Preprocessing Pipelines

StandardScaler is applied to Linear Regression and Ridge within their pipelines.
Tree-based models are used without scaling.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)

models = {
    "Linear Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LinearRegression())
    ]),

    "Ridge": Pipeline([
        ("scaler", StandardScaler()),
        ("model", Ridge(alpha=1.0))
    ]),

    "Random Forest": Pipeline([
        ("model", RandomForestRegressor(
            n_estimators=100,
            min_samples_leaf=2,
            random_state=42,
            n_jobs=1
        ))
    ]),

    "Gradient Boosting": Pipeline([
        ("model", GradientBoostingRegressor(
            n_estimators=100,
            learning_rate=0.1,
            max_depth=3,
            random_state=42
        ))
    ])
}

print("Models:", list(models))
```

## 3. Compare Models Using Grouped Cross-Validation

Five-fold GroupKFold keeps each engine entirely within either training
or validation in each fold. Models are compared using mean validation
MAE, RMSE, and R² across folds.

```python
from sklearn.model_selection import GroupKFold, cross_validate

cv = GroupKFold(n_splits=5)

scoring = {
    "MAE": "neg_mean_absolute_error",
    "RMSE": "neg_root_mean_squared_error",
    "R2": "r2"
}

results = []

for name, model in models.items():
    print(f"Evaluating: {name}")

    scores = cross_validate(
        model,
        X,
        y,
        groups=groups,
        cv=cv,
        scoring=scoring,
        n_jobs=-1,
        error_score="raise"
    )

    results.append({
        "Model": name,
        "MAE": -scores["test_MAE"].mean(),
        "RMSE": -scores["test_RMSE"].mean(),
        "R²": scores["test_R2"].mean()
    })

results_df = (
    pd.DataFrame(results)
    .sort_values("RMSE")
    .reset_index(drop=True)
)

display(results_df.round(3))
```

```python
from sklearn.model_selection import GridSearchCV

param_grids = {
    "Random Forest": {
        "model__n_estimators": [100],
        "model__max_depth": [10, None],
        "model__min_samples_leaf": [2, 5]
    },

    "Gradient Boosting": {
        "model__n_estimators": [100, 200],
        "model__learning_rate": [0.05, 0.1],
        "model__max_depth": [2, 3]
    }
}

searches = []

for name, param_grid in param_grids.items():
    print(f"Tuning: {name}")

    search = GridSearchCV(
        estimator=models[name],
        param_grid=param_grid,
        scoring=scoring,
        refit="RMSE",
        cv=cv,
        n_jobs=-1,
        error_score="raise"
    )

    search.fit(X, y, groups=groups)
    searches.append((name, search))

    print("Best parameters:", search.best_params_)
    print("Best CV RMSE:", round(-search.best_score_, 3))
```

### Tuning Results

Hyperparameter tuning slightly improved both models.
Gradient Boosting achieved the lowest mean cross-validation RMSE
of 37.236 cycles, compared with 37.617 for Random Forest.
The tuned Gradient Boosting model is the leading candidate,
pending evaluation on the official test set.

```python

columns = (
    ["engine_id", "cycle"]
    + [f"setting_{i}" for i in range(1, 4)]
    + [f"sensor_{i}" for i in range(1, 22)]
)


test_df = pd.read_csv(
    "../Data/Original/test_FD001.txt",
    sep=r"\s+",
    header=None,
    names=columns
)

y_test = pd.read_csv(
    "../Data/Original/RUL_FD001.txt",
    sep=r"\s+",
    header=None,
    names=["RUL"]
)["RUL"]


test_last = (
    test_df.sort_values(["engine_id", "cycle"])
    .groupby("engine_id")
    .tail(1)
    .sort_values("engine_id")
    .reset_index(drop=True)
)


X_test = test_last[X.columns].copy()

assert len(X_test) == len(y_test)

print("Test features:", X_test.shape)
print("Test target:", y_test.shape)

display(test_last[["engine_id", "cycle"]].head())
display(y_test.head())
```

```python
from sklearn.base import clone
from sklearn.metrics import (
    mean_absolute_error,
    root_mean_squared_error,
    r2_score
)
final_models = {
    name: clone(model)
    for name, model in models.items()
}

# for name, search in searches:
#     final_models[name] = clone(search.best_estimator_)

# test_results = []
# test_predictions = {}

# for name, model in final_models.items():

#     model.fit(X, y)

  
#     predictions = model.predict(X_test)
#     test_predictions[name] = predictions

#     test_results.append({
#         "Model": name,
#         "MAE": mean_absolute_error(y_test, predictions),
#         "RMSE": root_mean_squared_error(y_test, predictions),
#         "R²": r2_score(y_test, predictions)
#     })
for model in final_models.values():
    model.fit(X, y)

test_predictions = predict_rul(final_models, X_test)

test_results = []
for name, predictions in test_predictions.items():
    test_results.append({
        "Model": name,
        "MAE": mean_absolute_error(y_test, predictions),
        "RMSE": root_mean_squared_error(y_test, predictions),
        "R²": r2_score(y_test, predictions),
    })
test_results_df = (
    pd.DataFrame(test_results)
    .sort_values("RMSE")
    .reset_index(drop=True)
)

display(test_results_df.round(3))
```

```python
from pathlib import Path
import joblib
import json

models_dir = Path("../Models")
models_dir.mkdir(parents=True, exist_ok=True)

for name, model in final_models.items():
    filename = name.lower().replace(" ", "_") + ".joblib"
    joblib.dump(model, models_dir / filename)


with open(models_dir / "feature_columns.json", "w") as file:
    json.dump(X.columns.tolist(), file, indent=2)


test_results_df.to_csv(
    models_dir / "test_results.csv",
    index=False
)

print("Saved all four models, feature names, and test results.")
```

Predictions generally follow the ideal diagonal, but noticeable errors remain.
The plot suggests greater prediction spread at medium and high actual RUL.
Some engines with low actual RUL receive overestimated predictions,

## Experiment: Full RUL vs. RUL Capped at 125 Cycles

We compared two versions of each model. The first was trained and evaluated
using the original Remaining Useful Life (RUL) values. The second was trained
and evaluated with RUL capped at 125 cycles.

Both experiments used the same test engines. The results are reported
side by side for two different prediction targets.

```python
capped_results = []

for name, predictions in capped_predictions.items():
    predictions_at_125 = np.minimum(predictions, RUL_CAP)

    capped_results.append({
        "Model": name,
        "R²": r2_score(y_test_capped, predictions_at_125),
        "MAE": mean_absolute_error(y_test_capped, predictions_at_125),
        "RMSE": root_mean_squared_error(y_test_capped, predictions_at_125),
    })

comparison_df = test_results_df.merge(
    pd.DataFrame(capped_results),
    on="Model",
    suffixes=(" (Full RUL)", " (Capped at 125)"),
)

display(comparison_df[[
    "Model",
    "R² (Full RUL)", "R² (Capped at 125)",
    "MAE (Full RUL)", "MAE (Capped at 125)",
    "RMSE (Full RUL)", "RMSE (Capped at 125)",
]].round(3))
```
